## Practicing Basic Cleaning and Wrangling

```python
df[df['score_rank'].isna() == False]
```

The 5 records with **score_rank** having a numerical value also have a **userscore** that is greater than 0. Compared to the previous data sample, all records with **score_rank** of NaN have a **userscore** of 0.  

Additionally, the relationship between **score_rank**, **userscore**, **positive**, and **negative** seem illogical at initial glance. Record 2122 and Record 9662 both share a userscore value of 95, suggesting why they both share a **score_rank** of 100.0. This also presents that a higher **score_rank** reflects positively rather than a lower value, so this field may be measured in a percentile-sense. Renaming the column for clarity may be necessary. 

I have suspicion of the scoring, but I want to see if we can rebuild that column. Let's make a dataframe of the first several fields and include just the records with numerical **score_rank** values.

```python
df_scored = df[df['score_rank'].notna()][['appid', 'name', 'score_rank', 'positive', 'negative', 'userscore']]
df_scored['positive_ratio'] = (df['positive'] / (df['positive'] + df['negative'])) * 100

df_scored.head()
```

This confirms that this field's values come from a simple ratio, but it is of note that this data has rounded down all values. I'm not sure why that would be the case. With this information, we should be able to calculate and fill all remaining missing values for the **userscore** field as well as fill and sort the **score_rank** field.

```python
df['positive_ratio'] = ((df['positive'] / (df['positive'] + df['negative'])) * 100).round(0).astype('Int64')
df['positive_ratio'].head(15)
```

```python
df.head()
```

The **positive_ratio** field has been added to the end. Now we just need to reorder the field positions to group the relevant data together.

```python
cols = df.columns.tolist()
cols.remove('positive_ratio')
cols.insert(8, 'positive_ratio')
df = df[cols]
df.head()
```

```python
print(df['userscore'].value_counts()[0])
print(df['positive_ratio'].value_counts()[0])
```

The 0 values have been reduced from 9,995 counts in our 10,000 records within the **userscore** field to just 25 counts within the **positive_ratio** field, a substantial improvement over the source-available data. The **userscore** column should now be droppable.

```python
df.drop(columns=['userscore'], inplace=True)
df
```

## Week 2: EDA and Visualizations

### First, the issue with the excessive NaN values within **score_rank** should be addressed.

This has potential to be very insightful information on determining the effect of user satisfaction on popularity, and it would be lost value to drop the entire field. Online research has led to the finding of the .rank() method, which should allow for filling in the values with meaningful data while also shifting away from a percentile ranking.

```python
df['score_rank'] = df['positive_ratio'].rank(ascending=False, method='min')
df.head()
```

```python
df.sort_values('score_rank').tail(15)
```

```python
df.isna().sum()
```

(_This code above was initially ran before replacing the NaN values in **score_rank** and **positive_ratio**_.)  
The **score_rank** field now meaningful data in a large majority or records, aligning with the **postivie_ratio** field and only containing 12 NAN records. All NAN values in **score_rank** are due to a divide by 0 error caused by a total of 0 user reviews. We can replace them with a number outside of our range to specially mark them. Since the games are being ranked 1 through 10,000, we can replace NaNs in **score_rank** with 0, and we can replace with -1 in the **positive_ratio** field.

```python
df['positive_ratio'] = df['positive_ratio'].fillna(-1)
df['score_rank'] = df['score_rank'].fillna(0)
# df.sort_values('score_rank').tail(15)
df.sort_values('score_rank').head(15)
```

Now that they are filled with a more manageable value, let's group just the games that have been reviewed and will have a true ranking.

```python
df_ranked_reviewed = df[df['score_rank'] > 0]
df_ranked_reviewed
```

```python
df['developer'] = df['developer'].fillna('Unknown')
df['publisher'] = df['publisher'].fillna('Unknown')
df.isna().sum()
```

All NaN value have been removed, visualization can now begin.

### Data Visualization

```python
import matplotlib.pyplot as plt
```

_Histogram_ - **positive_ratio** distribution

```python
reviewed = df[df['positive_ratio'] >= 0]['positive_ratio']
unreviewed_count = (df['positive_ratio'] == -1).sum()

fig, ax = plt.subplots()
ax.hist(reviewed, bins=20)
ax.bar(-5, unreviewed_count, width=4, label='Not Reviewed', color='red')

plt.title('Distribution of Positive Ratio')
plt.xlabel('Positive Ratio')
plt.ylabel('Count')
plt.legend()
plt.show()
```

The distribution of positive_ratio across 10,000 titles was examined to understand how sentiment is spread across the dataset. A histogram was selected to visualize frequency across a continuous numeric range.  
The majority of reviewed games fall above a 75% positive ratio, with a separate concentration at -1 representing unreviewed titles and colored in red.

_Bar Chart_ - **publishers** value count

```python
top_publishers_count = df['publisher'].value_counts().head(10)
plt.bar(top_publishers_count.index, top_publishers_count.values)
plt.title('Top 10 Publishers by Game Count')
plt.xlabel('Publisher')
plt.ylabel('Number of Games')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

Game counts were aggregated by publisher to identify which publishers hold the largest presence in the dataset. A bar chart was appropriate for comparing individual category totals against one another. Ubisoft, EA, and SEGA lead by volume, each with over 100 titles represented each.

_Scatterplot_ - **price** vs **positive_ratio** 

```python
plt.scatter(df['positive_ratio'], df['price'] / 100, alpha=0.3)
plt.title('Positive Ratio vs Price')
plt.xlabel('Positive Ratio')
plt.ylabel('Price (USD)')
plt.show()
```

The relationship between price and positive_ratio was plotted to determine whether pricing correlates with player sentiment. A scatterplot was chosen to reveal any linear or clustered patterns between two numeric variables. No strong correlation is apparent, with sentiment varying widely across all price points.  
However, this plot does bring to question how price affects the likelihood of a game being reviewed. It would stand to reason that a lower priced game would be experienced by a higher number of players than a higher priced game, increasing to size of the total possible reviewers for the game. May be worth investigating further.

```python
import seaborn as sns
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, fmt='.2f', cmap='magma')
plt.title('Correlation Heatmap')
plt.show()
```
