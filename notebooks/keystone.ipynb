## Practicing Basic Cleaning and Wrangling

```python
df[df['score_rank'].isna() == False]
```

The 5 records with **score_rank** having a numerical value also have a **userscore** that is greater than 0. Compared to the previous data sample, all records with **score_rank** of NaN have a **userscore** of 0.  

Additionally, the relationship between **score_rank**, **userscore**, **positive**, and **negative** seem illogical at initial glance. Record 2122 and Record 9662 both share a userscore value of 95, suggesting why they both share a **score_rank** of 100.0. This also presents that a higher **score_rank** reflects positively rather than a lower value, so this field may be measured in a percentile-sense. Renaming the column for clarity may be necessary. 

I have suspicion of the scoring, but I want to see if we can rebuild that column. Let's make a dataframe of the first several fields and include just the records with numerical **score_rank** values.

```python
df_scored = df[df['score_rank'].notna()][['appid', 'name', 'score_rank', 'positive', 'negative', 'userscore']]
df_scored['positive_ratio'] = (df['positive'] / (df['positive'] + df['negative'])) * 100

df_scored.head()
```

This confirms that this field's values come from a simple ratio, but it is of note that this data has rounded down all values. I'm not sure why that would be the case. With this information, we should be able to calculate and fill all remaining missing values for the **userscore** field as well as fill and sort the **score_rank** field.

```python
df['positive_ratio'] = ((df['positive'] / (df['positive'] + df['negative'])) * 100).round(0).astype('Int64')
df['positive_ratio'].head(15)
```

## Week 2: EDA and Visualizations

### First, the issue with the excessive NaN values within **score_rank** should be addressed.

This has potential to be very insightful information on determining the effect of user satisfaction on popularity, and it would be lost value to drop the entire field. Online research has led to the finding of the .rank() method, which should allow for filling in the values with meaningful data while also shifting away from a percentile ranking.

```python
df['score_rank'] = df['positive_ratio'].rank(ascending=False, method='min')
df.head()
```

```python
df.sort_values('score_rank').tail(15)
```

```python
df.isna().sum()
```

(_This code above was initially ran before replacing the NaN values in **score_rank** and **positive_ratio**_.)  
The **score_rank** field now meaningful data in a large majority or records, aligning with the **postivie_ratio** field and only containing 12 NAN records. All NAN values in **score_rank** are due to a divide by 0 error caused by a total of 0 user reviews. We can replace them with a number outside of our range to specially mark them. Since the games are being ranked 1 through 10,000, we can replace NaNs in **score_rank** with 0, and we can replace with -1 in the **positive_ratio** field.

```python
df['positive_ratio'] = df['positive_ratio'].fillna(-1)
df['score_rank'] = df['score_rank'].fillna(0)
# df.sort_values('score_rank').tail(15)
df.sort_values('score_rank').head(15)
```

Now that they are filled with a more manageable value, let's group just the games that have been reviewed and will have a true ranking.

```python
df_ranked_reviewed = df[df['score_rank'] > 0]
df_ranked_reviewed
```

```python
df['developer'] = df['developer'].fillna('Unknown')
df['publisher'] = df['publisher'].fillna('Unknown')
df.isna().sum()
```

## Data Ethics Assessment

Before producing additional visualizations, the following questions were considered:

- **Bias in data collection:** There are several factors regarding data collection that are unknown. The exact source, collection methodology, collection date, and sampling method are all unclear. With that full data, there may be biases more explicit, but it can't be said with certainty otherwise.
- **Underrepresented groups:** Without having more details regarding the collection method and any potential biases, it is unclear if there are any categorical representation issues.
- **Misleading charts:** Early scatter plots used raw **owners_lower** values without acknowledging the nature of the ownership tiers, as each tier roughly doubled from 1x to 2x, 2x to 5x, then 5x to 10x, and so on. This could imply a linear ownership scale that doesn't exist. Revised visuals use explicit tick labels to address this and improve readability of axis tick values.
- **Telling the full story:** The relationship between **user_score** and **score_rank** initially relied on percentile ranking rather than ascending numerical ranking (specifically, rank 100 was a higher ranking than rank 1). This was documented in the heatmap analysis to prevent misinterpretation of the -0.92 correlation as a meaningful external finding, and the ranking was changed to represent the highest scores of 100 with a ranking of 1, shared among multiple entries before proceeding.
- **Limitations:** **user_score** treats a game with 5 reviews equally to one with 50,000. Review volume is not yet weighted, and any conclusions about sentiment should be understood in that context. Additionally, **owners** values are determined based on large buckets of ranges that double or more per step, requiring adjustment of the data reference for comparisons.

```python
fig, ax = plt.subplots(figsize=(10,6))

reviewed = df[df['user_score'] >= 0]['user_score']
unreviewed_count = (df['user_score'] == -1).sum()

ax.hist(reviewed, bins=25, range=(0, 100), color='#4C9BE8', edgecolor='white', linewidth=0.5, label='Reviewed games')
ax.set_xlim(0, 100)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.title('Distribution of User Satisfaction Score\nAmong Reviewed Steam Games', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('User Score (% Positive Reviews)', fontsize=11)
plt.ylabel('Number of Games', fontsize=11)

fig.text(0.5, -0.03,
    f'Note: {unreviewed_count} unreviewed games are excluded from this chart. '
    'User score = positive reviews / total reviews × 100.',
    ha='center', fontsize=9, color='gray')

plt.tight_layout()
plt.savefig('../visualizations/user_score_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Chart: User Score Distribution**

*What question does this answer?* What is the distribution of user satisfaction scores across Steam games?

*Why this chart type?* A histogram demonstrates the shape of a distribution, indicating whether the cluster is top-heavy, bottom-heavy, or evenly distributed.

*Design choices:* 25 bins across a 0–100 range gives easily readable incremenets without forcing the viewer to spend too much time deciphering which value is being represented. The bin for non-reviewed games was removed for visual clutter and possible confusion around coloration and placing below 0 on the x-axis, but they are noted in an annotation.

*Accuracy / integrity:* User score is calculated as positive reviews divided by total reviews, so it reflects proportion rather than volume. A game with 5 reviews and a perfect score is measured equally to a game with 50,000 reviews. This is a known limitation of the metric.

*Finding:* Score distribution is found heavily in the high end. Most reviewed games sit above 70%, suggesting that players who take the time to submit a review tend to be satisfied, or that dissatisfied players abandon a game without reviewing it.

```python
df['price_usd'] = df['last_recorded_price'] / 100

df_reviewed = df[df['user_score'] >= 0].copy()

df_reviewed['price_tier'] = 'Unknown'
df_reviewed.loc[df_reviewed['price_usd'] == 0, 'price_tier'] = 'Free'
df_reviewed.loc[(df_reviewed['price_usd'] > 0) & (df_reviewed['price_usd'] <= 4.99), 'price_tier'] = '$0.01-$4.99'
df_reviewed.loc[(df_reviewed['price_usd'] > 4.99) & (df_reviewed['price_usd'] <= 9.99), 'price_tier'] = '$5-$9.99'
df_reviewed.loc[(df_reviewed['price_usd'] > 9.99) & (df_reviewed['price_usd'] <= 19.99), 'price_tier'] = '$10-$19.99'
df_reviewed.loc[(df_reviewed['price_usd'] > 19.99) & (df_reviewed['price_usd'] <= 29.99), 'price_tier'] = '$20-$29.99'
df_reviewed.loc[(df_reviewed['price_usd'] > 29.99) & (df_reviewed['price_usd'] <= 39.99), 'price_tier'] = '$30-$39.99'
df_reviewed.loc[(df_reviewed['price_usd'] > 39.99) & (df_reviewed['price_usd'] <= 49.99), 'price_tier'] = '$40-$49.99'
df_reviewed.loc[(df_reviewed['price_usd'] > 49.99) & (df_reviewed['price_usd'] <= 59.99), 'price_tier'] = '$50-$59.99'
df_reviewed.loc[df_reviewed['price_usd'] > 59.99, 'price_tier'] = '$60+'

tiers = ['Free', '$0.01-$4.99', '$5-$9.99', '$10-$19.99', '$20-$29.99', '$30-$39.99', '$40-$49.99', '$50-$59.99', '$60+']

tier_summary = df_reviewed.groupby('price_tier')['user_score'].agg(['mean']).reindex(tiers)

fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(tier_summary.index, tier_summary['mean'], color='#4C9BE8', edgecolor='white')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.set_title('Average User Score by Price Tier\n(Reviewed Games Only)', fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Price Tier', fontsize=11)
ax.set_ylabel('Average User Score (%)', fontsize=11)
ax.set_ylim(0, 100)

plt.tight_layout()
plt.savefig('../visualizations/price_tier_vs_user_score.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Chart: Average User Score Across Price Tiers**

*What question does this answer?* Does price point correlate with user satisfaction?

*Why this chart type?* A bar chart is appropriate for comparing an average metric across designated categories. A scatter plot of unregulated values failed to indicate any meaningful value.

*Design choices:* Price values were converted from cents to dollars for readability and bucketed into nine tiers using primarily $10 increments to better capture differences across the mid-range market. Top and right chart borders were removed to reduce visual clutter. Unreviewed games were excluded.

*Accuracy / integrity:* The y-axis starts at 0. The dataset reflects last-recorded price only, so discounts or price changes over time are not captured.

*Finding:* There is no strong relationship between price and user satisfaction. Scores remain rather consistent across most tiers, suggesting that price point alone is not a reliable measurement of how well a game is received by players.
